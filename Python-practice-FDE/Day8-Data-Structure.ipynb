### 1.1 String Indexing
#### Indexing:   We can call individual letters from word
## Positive index start from 0 (left) ====>
## Negative index start from -1 (right) <======

P   Y    T    H    O    N
0   1    2    3    4    5
-6  -5  -4   -3   -2   -1 


```python
## Positive index
word = 'TECHVIUM'
print(word[0])
print(word[3])
print(word[7])
```

```python
## Negative index
word = 'TECHVIUM'
print(word[-1])
print(word[-5])
print(word[-7])
```

```python
word = 'PYTHON'
print('Last character', word[-6])
```

```python
fruits = ['orange','grape', 'banana']
print(fruits[2])
print(fruits[-3])
```

## String Slicing
### Slicing means extracting a portion of the string

### start index - where slicing begins, included
### stop index - where slicing ends, excluded
### step number - of positions to shift/jump  (it defines the direction to shift(jump))
### positive step - start should be less than end (left to right) incremental slicing
### negative step - start should be greater than stop (right to left) decremental slicing

### Example:  [start, stop, step]
p     y     t     h     o     n


```python
word = 'PYTHON'
print(word[3:5])
# start is 3
# stop is 4
```

```python
word = 'PYTHON'
print(word[4:3])
# start is 5
# stop is 3
```

```python
word = 'PYTHON'
print(word[3:3])
# start is 5
# stop is 3
```

```python
word = 'PYTHON'
print(word[2:7])
# start is 2
# stop is 6
### Python slicing will not give any error if the stop range is out of the range and also start point larger than end point
```

```python
## omiting the start index 
word = 'PYTHON'
print(word[:3])
print(word[:4])
print(word[:10])
```

```python
## omiting the stop index 
word = 'PYTHON'
print(word[3:])
print(word[4:])
print(word[10:]) # will not print
print(word[:])
```

```python
## Negative slicing process
word = 'PYTHON'
print(word[-4:3])  # --->
print(word[-4:2]) # --->  <--- (both)  ; o/p no response
print(word[-4:-5]) # no response
```

```python
## Negative slicing process
word = 'PYTHON'
print(word[-4:-5:-1]) # o/p-> T
```

```python
## Negative slicing process
word = 'PYTHON'
print(word[-4:-6:-1]) # o/p-> T
```

```python
word = 'veerafde'
print(word[3:5:1])
print(word[3:5:2])
print(word[3:5:3])
print(word[1:6:2])
```
