### 3. elif, else can not stand alone
### Correct Indentation example

```python
#Simple if
age = 20
if age >= 18:
    print("You are eligible to Vote.."); print("Adult..")
else:
    print ("You are NOT eligible to vote")
```

```python
#Passing value dynamic using input()
#By default input() function conside string
age = int(input("Enter Ur Age.."))
if age >= 18:
    print("You are eligible to Vote.."); print("Adult..")
else:
    print ("You are NOT eligible to vote")
```

```python
marks = 75
if marks >= 90:
    print("Grade A")
elif marks >=75:
    print ("Grade B")
elif marks >= 65:
    print("Grade C")
else:
    print("Fail...")

## Execute ONLY one matching block 
```

## Nested Conditions
### Condition(s) inside the condition

```python
# Nested condition
marks=70
lab = 60
if marks >= 60:
    print("Pass")
    if lab >= 80:  # nested if block
        print('Eligible for 1st Grade')
    else:
        print("Not eligible for 1st Grade")
else:
    print("Failed")
    
```
