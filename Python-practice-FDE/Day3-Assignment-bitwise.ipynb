## Left shift Right shift

```python
### Left Shift
x=15
x<<=1  #same as  x = x << 1
print(x)

# Before binary 1111   8+4+2+1
# After binary 11110   16+8+4+2+0
# left shift (<<) : move bits left and add bit position on right 0
```

```python
x = 19 # 10011
x<<=2 # 10011000
print(x)
```

```python
# Try it
x =21
x<<=3
print(x)
```

```python
### Right Shift
x=15   # 1111
x>>=1 # 0111
print (x)

# >>1
# before x=15 1111
# after 0111
```

```python
### Right Shift
x=15   # 1111
x>>=2 # 0111
print (x)

# >>2
# before x=15 1111
# after 0011

## right shift(>>) move right and discard right most bits
```

```python
# left shift
15 1111
15 << 1  11110
15 << 2  111100
15 << 3  1111000
# right shift
15   1111
15 >> 1   0111
15 >> 2   0011
15 >> 3   0001
```
