## Python lambda function
### lambda function is small anonymous function (a function without formal name)

```python
# Syntax lambda function

lambda arguments : expression
```

```python
# normal function
def add(a,b):
    return (a+b)
result = add(10,20)
print(result)
```

```python
# lambda format function
add = lambda a,b : a+b  #here a,b are parameter and a+b is expression
result = add(10,20) #  arguments 10,20
print(result)

# add variable name to strore the lambda function
# lambda is a keyword creating anonymous function
# a,b are parameters
# a+b is expression

##NOTE: For lambda not required any return keyword because lambda automatically return expression result
```

```python
mayur = lambda x : x*x
print(mayur(5))
```

```python
check = lambda x: "even" if x % 2 == 0 else "odd"
print(check(10))
print(check(3))
```

```python
check = lambda x: "Positive" if x > 0 == 0 else "Negative"
print(check(5))
print(check(-5))
```

### Walrus operators
#### walrus operator := is used to value to variable use same variable i same expression

```python
# without walrus operator
numbers = [10,20,30,40]
count = len(numbers)
if count > 3:
    print(count)
    
```

```python
# with walrus operator
numbers = [10,20,30,40]
if count := len(numbers) > 3:
    print(count)
```

```python
if (num := eval(input("enter value.."))) > 10:
    print("Number is greater than 10")
else: 
    print("number is less than 10")
```

```python
lambda x : x > 10
if (num := eval(input("enter number.."))):
    print("Number is greater than 10")
else: 
    print("number is less than 10")
```
