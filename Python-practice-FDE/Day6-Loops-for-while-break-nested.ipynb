## Loops
### for, while,break,nested for loop,for loop if combination, for loop different types, operators,for loop with range

```python
# for loop with string example
a = "Techvium"
for i in a:
    print(i)
```

```python
# for loop with list/array example
cloud_name = ['AWS', 'Azure', 'GCP']
for i in cloud_name:
    print(i)
    
```

```python
# for loop with if condition
a = [1,2,3,4,5]
for i in a:
    if i == 3:
        print(i)
        
```

```python
# for loop with if condition and break
a = [1,2,3,4,5]
for i in a:
    if i == 2:
        #print(i)
        break
    print(i)
```

```python
# for loop with dictionry
user = {
    'name' : 'Mayur',
    'course' : 'FDE',
    'company' : 'Microsoft'
}
for i in user:
    print(i)
# for loop calling total items
for i in user.items():
    print(i)
    #print(type(i))
# for loop calling values
for i in user.values():
    print(i)
# for loop calling keys
for i in user.keys():
    print(i)
    
```

```python
# for loop with range
total = 1
dev = range(1,5)
for i in dev:
    print('i:',i)
    total += i  # same as total = total + i
    print('tot.:', total)
print('Total: ',total)    
```

```python
x = [1,2,3,4,5,6,7]
for i in x:
    if i%2 ==1:
        print(i, '- Odd number')
    else:
        print("Even number - ", i)

## flow executions
# x - variable
# = - assignment operator
# [1,2,3,4,5,6,7] - list
# for  - loop
# i - variable 
# in - membership operator
# x - list
#if - condition

```
