## Python Exception Handling
### try, exception, else, finally, raise
#### alone try or exception block is not allowed

```python
a = 10
b = 0 # b=2
print(a/b)
```

```python
try:
    a = 10
    b = 0
    print(a/b)
except:
    print("Can not divide by zero..")
```

```python
## Showing exact excetion error (using Exception as e)
try:
    a = 10
    b = 0
    print(a/b)
except Exception as e:
    print(e)
    # e is variable that store exception object [error information]
```

```python
count = 1
while count <= 3:
    try:
        a = 10
        b = 1
        print(a/b)
        count +=1
    except Exception as e:
        print(e)
        count +=1

    
```

```python
try:
    a = 10
    b = 0
    #b = 2
    print(a/b)
    count +=1
except:
    print("Can not divide by zero..")
finally:
    print("Excecution completed..")
```

```python
age =10
if age < 18:
    raise ValueError("Age must be 18 or above...")
print("Eligible to vote")

## raise  - manually raise/generate exception
## ValueError - Predefined exception
## Age must be 18 or above --- Custom error message
```

```python
try:
    a = int(input("Enter number for a"))  # a = 'abcd' and b = 2313; o/p => ValueError
    b = input("Enter number for b")
    result = a/b
    print(result)
except ZeroDivisionError:
    print("can not divide by zero")
except ValueError:
    print("Please enter correct number only")
except TypeError:
    print("Invalid data type..")
except Exception as e:
    print("unexpected error: ", e)
finally:
    print("program execution completed..")
# a = 'abcd' and b = 2313; o/p => ValueError
# a = 99 and b = 'abcd'; o/p => TypeError
# a = 99 and b = 0; o/p => ZeroDivisionError
```

```python
try:
    a = int(input("Enter number for a"))  # a = 'abcd' and b = 2313; o/p => ValueError
    b = int(input("Enter number for b"))
    result = a/b
    print(result)
except (ZeroDivisionError, ValueError) as e:
    print("Please enter correct number only",e)
else: 
    print("else execution")
finally:
    print("program execution completed..")
```
