## 1.3 Python function : with parameter and return value
### Accept input argument from the function call
### Return keyword give result
### This return keyword we can pass input to another function

```python
def mayur(a, b): # here a and b are parameters
    return a+b  # return keyword give result
mayur(10,20)  # Here 10, 20 input argument to function call
    
```

```python
def calculate_bill(price, qty, discount): # price 1000, #qty 3, #discount 10
    total = price * qty
    disc_amt = total * discount / 100
    final_amt = total - disc_amt
    return final_amt  # i need final amount as a result
    #return total, disc_amt, final_amt ## Return multiple values as Tuple
calculate_bill(1000, 3, 10)
#print(calculate_bill(1000, 3, 10))
final_bill = calculate_bill(1000, 3, 10)   # saving result in a variable...
print("Final Bill: ", final_bill)
```

```python
## Example:  multiple values as return
def calculate_bill(price, qty, discount): # price 1000, #qty 3, #discount 10
    total = price * qty
    disc_amt = total * discount / 100
    final_amt = total - disc_amt
    return total, disc_amt, final_amt ## This will retunr as a Tuple
calculate_bill(1000, 3, 10)
```

#### Function parameters with keyword arguments
##### Allow argument to parameter any order without affecting result

```python
def mayur(a,b):
    return a-b
mayur(b = 10, a = 20)  ## This is keyword arguments
mayur(a = 20, b = 10)
# The order matters python assigned based on the position of values
#definetely i will change if you change position a and b
# We can overcome by using keyword arguments
```

```python
def student(name, age, course):
    return name, age, course
student(age = 36, name = 'Mayur', course = 'FDE') ## This keyword arguments
```

### Passing the result of one function to  another functions

```python
## Passing the result of one function to  another functions
def add_first_function(a, b):
    return a + b
result = add_first_function(3,2)

def mult_second_function(number):
    return number * 2
final_result = mult_second_function(result)

def div_third_function(number):
    return number / 3
actual_result = div_third_function(final_result)

print(actual_result)
```

## Function with local and global variables
### global variable - define outside of the function
### local variable - define inside of the function

```python
def mayur():
    name = "Mayur" ## Local variable - defined inside of the function
    print("Hello..", name)
mayur()
```

```python
name = "FDE"
def tvm():
    name = "Mayur" ## inside of the function local variable
    print("Hello ", name) ## print local variable
tvm()
print(name) ## print global variable
```

```python
sal = 200000
def update_salary():
    global sal ## this global block tells python dont create new local variable, use existing variable nameed sal
    sal = 250000
final_sal = update_salary()
print(sal)
# initialy salary is = 200000
# global block tells python to use the global variable so its change/update to 2500000
```

```python
def student(name, marks, details):
    print(name)
    print(marks)
    print(details)
student("Mayur", 80, details = 'Hyd')
```
